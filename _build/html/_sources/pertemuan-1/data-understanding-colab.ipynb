```python
import openeo
```

```python
connection = openeo.connect("openeo.dataspace.copernicus.eu").authenticate_oidc()
```

```python
# ============================================================
# AREA OF INTEREST (AOI)
# Polygon lokasi pengamatan di Kabupaten Gresik
# Format koordinat: [longitude (bujur), latitude (lintang)]
# ============================================================

aoi = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "properties": {},
      "geometry": {
        "type": "Polygon",
        "coordinates": [
          [
            [112.651501, -7.1533711],   # kiri-atas (NW)
            [112.6591246, -7.1533711],  # kanan-atas (NE)
            [112.6591246, -7.1615405],  # kanan-bawah (SE)
            [112.6516026, -7.1615405],  # kiri-bawah (SW)
            [112.651501, -7.1533711]    # kembali ke titik awal (menutup)
          ]
        ]
      }
    }
  ]
}
```

## BATCH JOB 1: NO2 (Nitrogen Dioksida)

```python
# ============================================================
# BATCH JOB: NO2
# Memuat data NO2 dari Sentinel-5P untuk area Gresik
#
# spatial_extent = bounding box area yang dianalisis:
#   - west  = longitude terkecil (batas kiri)
#   - south = latitude terkecil  (batas bawah)
#   - east  = longitude terbesar (batas kanan)
#   - north = latitude terbesar  (batas atas)
# CATATAN: south HARUS lebih kecil dari north (nilai lat negatif)
# ============================================================

s5_no2 = connection.load_collection(
    "SENTINEL_5P_L2",
    temporal_extent=["2025-08-24", "2026-08-24"],
    spatial_extent={"west": 112.651501,
                    "south": -7.1615405,
                    "east": 112.6591246,
                    "north": -7.1533711},
    bands=["NO2"],
)
```

```python
# ============================================================
# AGREGASI NO2
# 1. aggregate_temporal_period: rata-rata harian (menghindari data ganda)
# 2. aggregate_spatial:         rata-rata data di dalam polygon AOI
# ============================================================

s5_no2 = s5_no2.aggregate_temporal_period(reducer="mean", period="day")
s5_no2 = s5_no2.aggregate_spatial(reducer="mean", geometries=aoi)
```

```python
# Jalankan batch job NO2 di server openEO, hasil disimpan ke file .nc
job = s5_no2.execute_batch(title="NO2 Gresik", outputfile="no2_polutan_gresik.nc")
```
