# Mengunduh Data CO di Kabupaten Gresik

Notebook ini mengambil data konsentrasi **CO (Karbon Monoksida)** dari Sentinel-5P melalui openEO, lalu menyimpannya sebagai file netCDF (`.nc`) dan CSV (`.csv`).

```python
import openeo
```

```python
connection = openeo.connect("openeo.dataspace.copernicus.eu").authenticate_oidc()
```

## Area of Interest (AOI)

Polygon lokasi pengamatan di Kabupaten Gresik. Format koordinat: `[longitude, latitude]`.

```python
aoi = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "properties": {},
      "geometry": {
        "type": "Polygon",
        "coordinates": [[
          [112.6193968, -7.1514786],
          [112.6600158, -7.1514786],
          [112.6600158, -7.1927923],
          [112.619805,  -7.1927923],
          [112.6193968, -7.1514786]
        ]]
      }
    }
  ]
}
```

## Load Data CO

```python
s5 = connection.load_collection(
    "SENTINEL_5P_L2",
    temporal_extent=["2025-08-24", "2026-08-24"],
    spatial_extent={
        "west": 112.6193968,
        "south": -7.1927923,
        "east": 112.6600158,
        "north": -7.1514786,
    },
    bands=["CO"],
)
```

```python
# Rata-rata harian, lalu rata-rata di dalam polygon AOI
s5 = s5.aggregate_temporal_period(reducer="mean", period="day")
s5 = s5.aggregate_spatial(reducer="mean", geometries=aoi)
```

## Jalankan Batch Job

```python
job = s5.execute_batch(title="CO Gresik", outputfile="../data/nc/polutan_CO_gresik.nc")
```

## Konversi netCDF ke CSV

```python
import netCDF4
import pandas as pd

# Buka file NetCDF
ds = netCDF4.Dataset("../data/nc/polutan_CO_gresik.nc")

# Ambil data CO
co = ds.variables["CO"][0, :]

# Ambil waktu
time = ds.variables["t"][:]

# Ubah waktu menjadi tanggal
dates = netCDF4.num2date(
    time,
    units=ds.variables["t"].units
)

# Buat tanggal lengkap selama periode penelitian
full_dates = pd.date_range(
    start="2025-08-24",
    end="2026-08-24",
    freq="D"
)

# Pasangkan tanggal dengan nilai CO
co_data = {
    date.strftime("%Y-%m-%d"): float(value)
    for date, value in zip(dates, co)
}

# Buat tabel dengan semua tanggal
df = pd.DataFrame({
    "date": full_dates.strftime("%Y-%m-%d")
})

# Masukkan nilai CO yang tersedia
df["CO"] = df["date"].map(co_data)

# Simpan ke CSV
df.to_csv(
    "../data/csv/CO_gresik_timeseries.csv",
    index=False
)

print("CSV berhasil disimpan")
print("Jumlah baris:", len(df))
```
