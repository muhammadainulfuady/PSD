# Mengunduh Data SO2 di Kabupaten Gresik

Notebook ini mengambil data konsentrasi **SO2 (Sulfur Dioksida)** dari Sentinel-5P melalui openEO, lalu menyimpannya sebagai file netCDF (`.nc`) dan CSV (`.csv`).

```python
import openeo
```

```python
connection = openeo.connect("openeo.dataspace.copernicus.eu").authenticate_oidc()
```

## Area of Interest (AOI)

Polygon lokasi pengamatan di Kabupaten Gresik. Format koordinat: `[longitude, latitude]`.

```python
aoi = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "properties": {},
      "geometry": {
        "type": "Polygon",
        "coordinates": [[
          [112.6193968, -7.1514786],
          [112.6600158, -7.1514786],
          [112.6600158, -7.1927923],
          [112.619805,  -7.1927923],
          [112.6193968, -7.1514786]
        ]]
      }
    }
  ]
}
```

## Load Data SO2

```python
s5 = connection.load_collection(
    "SENTINEL_5P_L2",
    temporal_extent=["2025-08-24", "2026-08-24"],
    spatial_extent={
        "west": 112.6193968,
        "south": -7.1927923,
        "east": 112.6600158,
        "north": -7.1514786,
    },
    bands=["SO2"],
)
```

```python
# Rata-rata harian, lalu rata-rata di dalam polygon AOI
s5 = s5.aggregate_temporal_period(reducer="mean", period="day")
s5 = s5.aggregate_spatial(reducer="mean", geometries=aoi)
```

## Jalankan Batch Job

```python
job = s5.execute_batch(title="SO2 Gresik", outputfile="../data/nc/polutan_SO2_gresik.nc")
```

## Konversi netCDF ke CSV

```python
import netCDF4
import pandas as pd

ds = netCDF4.Dataset("../data/nc/polutan_SO2_gresik.nc")

so2 = ds.variables["SO2"][0, :]   # feature 0, semua waktu
time = ds.variables["t"][:]
time_units = ds.variables["t"].units
dates = netCDF4.num2date(time, units=time_units)

# Deret tanggal penuh 1 tahun
full_dates = pd.date_range(start="2025-08-24", end="2026-08-23", freq="D")
so2_map = {d.strftime("%Y-%m-%d"): float(v) for d, v in zip(dates, so2)}

df = pd.DataFrame({"date": full_dates.strftime("%Y-%m-%d")})
df["SO2"] = df["date"].map(so2_map)

df.to_csv("../data/csv/SO2_gresik_timeseries.csv", index=False)
print("CSV disimpan:", len(df), "baris")
```
