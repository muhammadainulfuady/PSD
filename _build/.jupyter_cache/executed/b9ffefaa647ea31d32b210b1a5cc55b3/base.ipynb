```python
import pandas as pd

# 1. Baca berkas CSV polutan gresik
csv_path = "../data/csv/polutan_gresik.csv"
df = pd.read_csv(csv_path)

# 2. Opsi A: Hapus baris yang ke-4 polutannya kosong sekaligus (hapus tanggal tanpa data polutan)
df_clean_any = df.dropna(subset=['CH4', 'CO', 'NO2', 'SO2'], how='all')

# 3. Opsi B: Hapus baris yang mengandung NaN sama sekali (hanya tanggal yang 4 polutannya lengkap terisi)
df_clean_full = df.dropna()

# 4. Simpan ke berkas Excel (.xlsx) dengan beberapa Sheet
output_excel = "polutan_gresik_clean.xlsx"
with pd.ExcelWriter(output_excel, engine="openpyxl") as writer:
    # Sheet 1: Data yang setidaknya memiliki 1 polutan terisi
    df_clean_any.to_excel(writer, sheet_name="Data_Terisi", index=False)
    
    # Sheet 2: Data yang 4 polutannya lengkap 100%
    df_clean_full.to_excel(writer, sheet_name="Lengkap_4_Polutan", index=False)
    
    # Sheet Terpisah per Polutan (Tanpa data kosong)
    df[['date', 'CH4']].dropna().to_excel(writer, sheet_name="CH4_Clean", index=False)
    df[['date', 'CO']].dropna().to_excel(writer, sheet_name="CO_Clean", index=False)
    df[['date', 'NO2']].dropna().to_excel(writer, sheet_name="NO2_Clean", index=False)
    df[['date', 'SO2']].dropna().to_excel(writer, sheet_name="SO2_Clean", index=False)

print(f"Berhasil mengekspor data ke {output_excel} tanpa data kosong!")
```
