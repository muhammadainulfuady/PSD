```python
import pandas as pd
import os
```

```python
# Path file CSV hasil crawling
path_ch4 = "../data/csv/CH4_gresik_timeseries.csv"
path_co  = "../data/csv/CO_gresik_timeseries.csv"
path_no2 = "../data/csv/NO2_gresik_timeseries.csv"
path_so2 = "../data/csv/SO2_gresik_timeseries.csv"

# Read CSV
df_ch4 = pd.read_csv(path_ch4)
df_co  = pd.read_csv(path_co)
df_no2 = pd.read_csv(path_no2)
df_so2 = pd.read_csv(path_so2)

print(f"Data CH4  : {df_ch4.shape[0]} baris, {df_ch4.shape[1]} kolom")
print(f"Data CO   : {df_co.shape[0]} baris, {df_co.shape[1]} kolom")
print(f"Data NO2  : {df_no2.shape[0]} baris, {df_no2.shape[1]} kolom")
print(f"Data SO2  : {df_so2.shape[0]} baris, {df_so2.shape[1]} kolom")
```

```python
# Merge berurutan berdasarkan kolom 'date'
df_merged = df_ch4.merge(df_co, on="date", how="outer")
df_merged = df_merged.merge(df_no2, on="date", how="outer")
df_merged = df_merged.merge(df_so2, on="date", how="outer")

# Urutkan tanggal secara kronologis
df_merged["date"] = pd.to_datetime(df_merged["date"])
df_merged = df_merged.sort_values("date").reset_index(drop=True)
df_merged["date"] = df_merged["date"].dt.strftime("%Y-%m-%d")

print(f"Hasil Penggabungan Terpadu: {df_merged.shape[0]} baris, {df_merged.shape[1]} kolom")
display(df_merged.head(10))
```

```python
print("=== Jumlah Missing Values (Sel Kosong) ===")
print(df_merged.isnull().sum())

print("\n=== Jumlah Data Valid (Terisi) ===")
print(df_merged.notnull().sum())

print("\n=== Persentase Data Valid (%) ===")
print((df_merged.notnull().mean() * 100).round(2).astype(str) + " %")
```

```python
output_csv_path = "../data/csv/polutan_gresik.csv"
os.makedirs("../data/csv", exist_ok=True)

# Simpan ke file CSV terpadu
df_merged.to_csv(output_csv_path, index=False)
print(f"[SUCCESS] Berhasil menyimpan dataset 4 polutan terpadu di: {output_csv_path}")
```
